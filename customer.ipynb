1. Missing value handle

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.mean()))
```

2. Column Name Rename

```python
df.columns
```

```python
df.columns = df.columns.str.lower()
```

```python
df.columns
```

```python
df.colums = df.columns.str.replace(" ","_")
```

```python
df = df.rename(columns={'purchase amount (usd)':'purchase_amount'})
```

3. Age group ready korbo

```python
df["age_group"] = pd.qcut(df['age'],q=4,labels=['Young','Middle','Imediate Senior','Senior'])
```
